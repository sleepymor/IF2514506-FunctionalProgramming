## 1. What is a Lambda Function?

A lambda expression (or lambda form) is used to create **anonymous functions** - functions without a name. 

In Python, a lambda function can take any number of arguments, but it can only have **one expression**.

**Syntax:**
```python
lambda [arguments] : [expression]
```

The expression is evaluated and returned automatically - there is no need for an explicit `return` statement.

### Basic Examples

**Example 1: Lambda with a single argument**

````python
# TODO: Create a lambda that adds 10 to a number
x = lambda a: a + 10
print(x(5))
````

**Example 2: Lambda with two arguments**

````python
# TODO: Create a lambda that multiplies two numbers
x = lambda a, b: a * b
print(x(5, 6))
````

**Example 3: Lambda with multiple arguments**

````python
# TODO: Lambda with 3 arguments
x = lambda a, b, c: a + b + c
print(x(1, 2, 3))
````

## 2. Why Use Lambda Functions?

The real power of lambda functions becomes apparent when you need a small, throwaway function for a short period of time. They are commonly used as **anonymous functions inside other functions** - particularly with higher-order functions like `map()`, `filter()`, and `sorted()`.

Lambdas are especially useful when:
- You need a simple function for a single use case
- Writing a full `def` statement feels verbose
- Working with functional programming patterns

````python
# TODO: Understand how lambda returns a function
def myfunc(n):
    return lambda a: a * n

mydoubler = myfunc(2)  # Returns a function that multiplies by 2
mytripler = myfunc(3)  # Returns a function that multiplies by 3

print(mydoubler(11))
print(mytripler(11))
````

> Notice how `mydoubler` and `mytripler` are functions created dynamically from `myfunc`. This is a common functional programming pattern!

## 3. Lambda with Built-in Functions

Lambdas work beautifully with Python's built-in functional tools.

### 3.1 Using `map()` with Lambda

`map(function, iterable)` applies the function to every item in the iterable.

```python
map(lambda x: x * 2, [1, 2, 3, 4])  # Returns doubled values
```

````python
# TODO: Use map() with lambda to double each number
numbers = [1, 2, 3, 4, 5]
doubled = list(map(lambda x: x * 2, numbers))
print(doubled)
````

### 3.2 Using `filter()` with Lambda

`filter(function, iterable)` returns items from the iterable for which the function returns `True`.

```python
filter(lambda x: x % 2 == 0, [1, 2, 3, 4])  # Returns even numbers
```

````python
# TODO: Use filter() with lambda to get even numbers
numbers = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)
````

### 3.3 Using `sorted()` with Lambda

`sorted()` can take a `key` parameter. Using lambda with `key` lets us customize sorting.

````python
# TODO: Sort a list of tuples by second element
items = [('apple', 5), ('banana', 2), ('cherry', 8)]
sorted_items = sorted(items, key=lambda x: x[1])
print(sorted_items)
````

````python
# TODO: Sort strings by length
words = ['cat', 'elephant', 'dog', 'bear']
sorted_words = sorted(words, key=lambda w: len(w))
print(sorted_words)
````

## 4. Limitations of Lambda Functions

While lambdas are convenient, they have important restrictions:

- **Single expression only**: You cannot include multiple statements (like `if`, `for`, `while` blocks) directly.
- **Limited readability**: Complex logic in a lambda is hard to read and debug.
- **No docstrings**: Lambdas cannot have documentation strings like regular functions.
- **No assignment**: You cannot assign variables inside a lambda expression.
- **Best for simple operations**: For anything non-trivial, use `def` instead.

### When to prefer `def` over `lambda`:
- When the function needs a name for clarity
- When logic is complex (multiple conditions, loops, etc.)
- When you need to test, debug, or document the function

### Comparison: `def` vs `lambda`

````python
# TODO: Compare def vs lambda
# Using def
def add(x, y):
    return x + y

# Using lambda
add_lambda = lambda x, y: x + y

print(add(2, 3))
print(add_lambda(2, 3))
````
