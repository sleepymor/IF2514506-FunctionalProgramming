## 1. What are Decorators?

A **decorator** is a design pattern in Python that allows you to modify or enhance the behavior of a function or method without changing its code. 

Decorators wrap a function, returning a new function (or callable) that adds extra functionality before, after, or around the original function's execution.

Decorators are built on top of:
- Functions as first-class objects
- Closures
- Higher-order functions

### Basic Idea

Instead of manually wrapping:
```python
def my_func(): ...
my_func = decorator(my_func)
```

We can use the `@` syntax:
```python
@decorator
def my_func(): ...
```

## 2. Creating a Basic Decorator

````python
# TODO: Basic decorator example
def greet_decorator(func):
    def wrapper():
        print("--- Before function call ---")
        func()
        print("--- After function call ---")
    return wrapper

@greet_decorator
def say_hello():
    print("Hello, World!")

say_hello()
````

## 5. Stacking Multiple Decorators

You can apply multiple decorators to a single function. They are applied from **bottom to top** (closest to function first), or equivalently the order is such that the top decorator wraps the result of the ones below.

```python
@dec1
@dec2
def func(): ...
```

This is equivalent to: `func = dec1(dec2(func))`

## 7. Common Use Cases for Decorators

Decorators are widely used in Python for:

- **Logging**: Track function calls and parameters
- **Timing/Performance**: Measure execution time
- **Authentication/Authorization**: Check user permissions
- **Caching/Memoization**: Store results of expensive calls (see `functools.lru_cache`)
- **Input validation**: Verify function arguments
- **Rate limiting**: Control how often functions can be called
- **Debugging**: Add debug information
- **Web frameworks**: Route handling (Flask, FastAPI use decorators heavily)

## 9. Summary

- **Decorators** wrap functions to modify their behavior without changing source code.
- The `@decorator` syntax is syntactic sugar for `func = decorator(func)`.
- Use `*args` and `**kwargs` to make decorators work with any function signature.
- Use `functools.wraps` to preserve original function metadata.
- Multiple decorators stack with the closest-to-function decorator applied first in the wrapping chain.
- Decorators can take arguments by adding an extra nesting layer.
- Decorators are essential for cross-cutting concerns like logging, timing, caching, and auth.
