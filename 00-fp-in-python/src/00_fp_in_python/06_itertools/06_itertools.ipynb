## 1. Introduction to Itertools

    The `itertools` module provides a collection of fast, memory-efficient tools for working with iterators. These functions are inspired by functional programming and APL/Haskell constructs.

    Key benefits:
    - **Memory efficient**: All functions return iterators (lazy evaluation)
    - **Composable**: Can be chained together
    - **Fast**: Implemented in C for performance
    - **Standard library**: No external dependencies

## 2. Infinite Iterators

    These create infinite sequences - use with care (slice with `islice` or break conditions).

### 2.1 `itertools.count(start=0, step=1)`

    Returns an infinite counter.

```python
# TODO: count examples
import itertools

# Count from 0
c = itertools.count()
for _ in range(5):
    print(next(c), end=' ')  # 0 1 2 3 4
print()

# Count from 10 with step 2
c2 = itertools.count(start=10, step=2)
for _ in range(5):
    print(next(c2), end=' ')  # 10 12 14 16 18
```

### 2.2 `itertools.cycle(iterable)`

    Cycles infinitely through the elements of an iterable.

```python
# TODO: cycle example
import itertools

colors = itertools.cycle(['red', 'green', 'blue'])
for _ in range(6):
    print(next(colors), end=' ')  # red green blue red green blue
```

### 2.3 `itertools.repeat(object[, times])`

    Repeats an object infinitely (or up to `times`).

```python
# TODO: repeat example
import itertools

# Repeat 5 times
for x in itertools.repeat('Hello', 3):
    print(x)

# Useful with map
squares = list(map(pow, range(1, 5), itertools.repeat(2)))
print(squares)  # [1, 4, 9, 16]
```

## 3. Finite Iterators - Combining & Chaining

### 3.1 `itertools.chain(*iterables)`

    Flattens multiple iterables into one sequence.

```python
# TODO: chain example
import itertools

result = list(itertools.chain([1, 2, 3], ['a', 'b'], (4, 5)))
print(result)  # [1, 2, 3, 'a', 'b', 4, 5]

# Also chain.from_iterable
nested = [[1,2], [3,4], [5]]
flat = list(itertools.chain.from_iterable(nested))
print(flat)  # [1, 2, 3, 4, 5]
```

### 3.2 `itertools.zip_longest(*iterables, fillvalue=None)`

    Like `zip()`, but continues until all iterables are exhausted. Fills missing values.

```python
# TODO: zip_longest example
import itertools

a = [1, 2, 3]
b = ['a', 'b']

# Regular zip truncates
print(list(zip(a, b)))  # [(1, 'a'), (2, 'b')]

# zip_longest fills
print(list(itertools.zip_longest(a, b, fillvalue='N/A')))  # [(1, 'a'), (2, 'b'), (3, 'N/A')]
```

## 4. Slicing and Filtering Iterators

### 4.1 `itertools.islice(iterable, start, stop[, step])`

    Slices an iterator without consuming it all into a list. Works like regular slicing but lazily.

```python
# TODO: islice example
import itertools

data = range(20)
# Get elements from index 5 to 10
sliced = list(itertools.islice(data, 5, 10))
print(sliced)  # [5, 6, 7, 8, 9]

# With step
sliced2 = list(itertools.islice(data, 0, 20, 3))
print(sliced2)  # [0, 3, 6, 9, 12, 15, 18]

# Slice infinite iterator
first_5 = list(itertools.islice(itertools.count(), 5))
print(first_5)
```

### 4.2 `itertools.takewhile(predicate, iterable)`

    Takes elements while predicate is True. Stops at first False.

```python
# TODO: takewhile example
import itertools

numbers = [1, 2, 3, 4, 0, 5, 6]
result = list(itertools.takewhile(lambda x: x < 5, numbers))
print(result)  # [1, 2, 3, 4] (stops at 0)
```

### 4.3 `itertools.dropwhile(predicate, iterable)`

    Drops elements while predicate is True. Returns remaining once False.

```python
# TODO: dropwhile example
import itertools

numbers = [1, 2, 3, 4, 0, 5, 6]
result = list(itertools.dropwhile(lambda x: x < 5, numbers))
print(result)  # [0, 5, 6]
```

## 5. Grouping and Accumulating

### 5.1 `itertools.groupby(iterable, key=None)`

    Groups consecutive elements that have the same key. **Important**: The iterable must be sorted by the key for meaningful grouping!

```python
# TODO: groupby example
import itertools

# Group by even/odd
numbers = [1, 1, 2, 2, 2, 3, 4, 4]
numbers.sort()  # Sort for proper grouping
for key, group in itertools.groupby(numbers):
    print(f"{key}: {list(group)}")
```

```python
# TODO: groupby with key function
import itertools

data = ['apple', 'apricot', 'banana', 'blueberry', 'cherry']
data.sort()  # Sort by first letter
for key, group in itertools.groupby(data, key=lambda x: x[0]):
    print(f"{key}: {list(group)}")
```

### 5.2 `itertools.accumulate(iterable, func=operator.add, *, initial=None)`

    Returns accumulated results. By default sums; can use custom function.

```python
# TODO: accumulate examples
import itertools
import operator

numbers = [1, 2, 3, 4, 5]
# Default sum
print(list(itertools.accumulate(numbers)))  # [1, 3, 6, 10, 15]

# Product
print(list(itertools.accumulate(numbers, operator.mul)))  # [1, 2, 6, 24, 120]

# Custom - max
print(list(itertools.accumulate(numbers, max)))  # [1, 2, 3, 4, 5] but also tracks max
```

```python
# TODO: accumulate with running differences
import itertools

data = [10, 20, 30, 40]
differences = list(itertools.accumulate(data, lambda a, b: b - a))
print(differences)
```

## 6. Combinatoric Generators

    These generate cartesian products, permutations, and combinations - perfect for combinatorics problems.

### 6.1 `itertools.product(*iterables, repeat=1)`

    Cartesian product of input iterables. Equivalent to nested for-loops.

```python
# TODO: product example
import itertools

# Product of two iterables
colors = ['red', 'blue']
sizes = ['S', 'M', 'L']
combos = list(itertools.product(colors, sizes))
print(combos)
# [('red', 'S'), ('red', 'M'), ('red', 'L'), ('blue', 'S'), ('blue', 'M'), ('blue', 'L')]

# With repeat (like dice rolls)
dice = list(itertools.product(range(1, 7), repeat=2))
print(f"Total pairs: {len(dice)}")
```

### 6.2 `itertools.permutations(iterable, r=None)`

    Returns all possible orderings (permutations) of length `r`. Order matters, no repeats of positions.

```python
# TODO: permutations example
import itertools

items = ['A', 'B', 'C']
# All permutations of all 3
perms = list(itertools.permutations(items))
print("All:", perms)

# Permutations of length 2
perms2 = list(itertools.permutations(items, r=2))
print("r=2:", perms2)
```

### 6.3 `itertools.combinations(iterable, r)`

    Returns all combinations of length `r`. Order does **not** matter.

```python
# TODO: combinations example
import itertools

items = ['A', 'B', 'C', 'D']
# Combinations of length 2
combs = list(itertools.combinations(items, r=2))
print(combs)
print(f"Count: {len(combs)}")
```

### 6.4 `itertools.combinations_with_replacement(iterable, r)`

    Like combinations, but elements can be repeated.

```python
# TODO: combinations_with_replacement
import itertools

items = ['A', 'B', 'C']
combs_rep = list(itertools.combinations_with_replacement(items, r=2))
print(combs_rep)
```

## 7. Practical Examples

### 7.1 Chunking Data

```python
# TODO: Chunk data with islice or custom approach
import itertools

def chunker(iterable, size):
    it = iter(iterable)
    while True:
        chunk = tuple(itertools.islice(it, size))
        if not chunk:
            break
        yield chunk

data = range(10)
for chunk in chunker(data, 3):
    print(chunk)
```

## 9. Summary

    - `itertools` provides **memory-efficient, lazy iterators** for functional-style iteration.
    - **Infinite iterators**: `count()`, `cycle()`, `repeat()` (use with slicing/bounds).
    - **Chaining**: `chain()` flattens iterables; `zip_longest()` handles uneven lengths.
    - **Filtering/Slicing**: `islice()`, `takewhile()`, `dropwhile()` work on iterators.
    - **Grouping**: `groupby()` groups consecutive items - sort by key first!
    - **Accumulation**: `accumulate()` builds running totals with optional custom functions.
    - **Combinatorics**: `product()` (cartesian), `permutations()` (ordered), `combinations()` (unordered), `combinations_with_replacement()`.
    - These tools help write **concise, efficient, and functional** code.
