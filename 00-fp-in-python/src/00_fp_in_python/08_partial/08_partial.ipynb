## 1. What is Partial Application?

    **Partial application** is the process of fixing a number of arguments to a function, producing another function with a smaller arity (fewer arguments). 

    It's different from **currying** (Haskell-style):
    - **Partial application**: Fix some args now, get a function that takes the rest (can take multiple remaining args)
    - **Currying**: Transform a function that takes multiple args into a sequence of unary (single-arg) functions

    In Python, `functools.partial()` gives us partial application.

## 9. Summary

    - **Partial application** fixes some function arguments to create a new function with fewer args.
    - `functools.partial()` provides a clean way to do this in Python.
    - Positional args fill left-to-right; keyword args can be pre-filled and overridden.
    - `partialmethod()` works similarly but for class methods.
    - Partial application is great for **callbacks, API clients, logging, and creating specialized functions**.
    - **Partial application vs currying**: Partial takes some args now (remaining can take multiple); currying breaks into unary functions.
    - You can always implement partial manually with closures - understanding the concept is key!
