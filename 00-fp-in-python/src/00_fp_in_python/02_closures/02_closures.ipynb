## 1. What is a Closure?

A **closure** is a nested function that remembers and can access variables from its enclosing (outer) function's scope, even after the outer function has finished executing.

For a closure to exist, we need:
1. A nested function (function inside another function)
2. The nested function must reference a variable from the outer function
3. The outer function must return the nested function

### Basic Example

```python
# TODO: Understand a simple closure
def outer(msg):
    def inner():
        print(f"Message: {msg}")
    return inner

my_func = outer("Hello, Closure!")
my_func()  # 'msg' is remembered even after outer() returned!
```

> Even though `outer()` has finished execution, the `inner()` function still has access to `msg`. This is the essence of a closure!

## 2. Nested Functions vs Closures

All closures are nested functions, but not all nested functions are closures. A nested function becomes a closure when it captures (encloses) variables from the outer scope and is returned or used outside.

```python
# TODO: Nested function (not necessarily a closure if not returned)
def outer():
    x = 10
    def inner():
        print(x)
    inner()  # Called inside outer - this is just nesting

outer()
```

The above is a nested function. To make it a closure, we return `inner` instead of calling it directly.

```python
# TODO: True closure - returned and called later
def outer():
    x = 10
    def inner():
        print(x)
    return inner

f = outer()
f()  # Called after outer returns
```

## 3. Creating Multiple Closures

Each time we call the outer function, a new closure is created with its own copy of the enclosed variables.

```python
# TODO: Multiple independent closures
def make_multiplier(n):
    def multiplier(x):
        return x * n
    return multiplier

times2 = make_multiplier(2)
times3 = make_multiplier(3)
times5 = make_multiplier(5)

print(times2(10))
print(times3(10))
print(times5(10))
```

## 4. Closures for Preserving State

Closures are great for maintaining state without using global variables or classes. They encapsulate data.

```python
# TODO: Closure-based counter
def counter():
    count = 0
    def increment():
        nonlocal count  # Needed to assign to count in nested scope
        count += 1
        return count
    return increment

c1 = counter()
print(c1())
print(c1())
print(c1())

c2 = counter()  # New counter, independent state
print(c2())
print(c1())  # c1's state is preserved
```

## 5. Understanding `nonlocal`

- Use `nonlocal` when you want to **assign** to a variable in the enclosing (but non-global) scope.
- If you're only **reading** the variable, `nonlocal` is not required.
- `nonlocal` is different from `global` - it only looks in enclosing function scopes, not at module level.

```python
# TODO: Reading without nonlocal (works fine)
def outer():
    x = 5
    def inner():
        print(x)  # Reading is okay
    return inner

outer()()
```

## 6. Practical Examples of Closures

Closures are used in many real-world scenarios, including decorators (which we'll see next), data encapsulation, and callback functions.

### 6.1 Custom Greeting Generator

```python
# TODO: Custom greeting closure
def make_greeting(name):
    def greet():
        return f"Hello, {name}! Welcome back."
    return greet

greet_alice = make_greeting("Alice")
greet_bob = make_greeting("Bob")
print(greet_alice())
print(greet_bob())
```

### 6.2 Simple Logger

```python
# TODO: Logger using closure
def logger(prefix):
    def log(message):
        print(f"[{prefix}] {message}")
    return log

info = logger("INFO")
error = logger("ERROR")
info("Application started")
error("Something went wrong")
```

### 6.3 Accumulator

```python
# TODO: Accumulator that adds values
def accumulator(start=0):
    total = start
    def add(num):
        nonlocal total
        total += num
        return total
    return add

acc = accumulator(10)
print(acc(5))
print(acc(3))
print(acc(7))
```

## 7. Inspecting Closures

In Python, we can inspect a closure's enclosed variables using the `__closure__` attribute.

```python
# TODO: Inspect closure attributes
def outer(msg):
    def inner():
        print(msg)
    return inner

f = outer("Inspect me")
print(f.__closure__)
print(f.__closure__[0].cell_contents)
print(f.__name__)
```

## 9. Summary

- A **closure** is a nested function that remembers variables from its enclosing scope even after the outer function returns.
- Closures enable **data hiding** and **state preservation** without global variables.
- Use `nonlocal` to assign to variables in the enclosing function's scope.
- Reading enclosed variables works without `nonlocal`.
- Each call to the outer function creates a **new independent closure**.
- Closures are fundamental to understanding **decorators** in Python.
