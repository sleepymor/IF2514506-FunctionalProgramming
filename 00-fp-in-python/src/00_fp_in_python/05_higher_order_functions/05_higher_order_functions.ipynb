## 1. What are Higher-Order Functions?

In functional programming, a **higher-order function (HOF)** is a function that does at least one of the following:

1. Takes one or more functions as arguments
    2. Returns a function as its result

Functions in Python are **first-class citizens**, meaning they can be:

- Assigned to variables
- Passed as arguments to other functions
- Returned from other functions
- Stored in data structures

### Functions as First-Class Objects

```python
# TODO: Functions are first-class
def greet(name):
    return f"Hello, {name}!"

# Assign to variable
say_hello = greet
print(say_hello("Alice"))

# Store in list
funcs = [greet, len, str]
print(funcs[0]("Bob"))
```

## 2. Functions that Take Functions as Arguments

```python
# TODO: Custom HOF that takes a function
def apply_twice(func, arg):
    return func(func(arg))

def add_five(x):
    return x + 5

print(apply_twice(add_five, 10))  # (10+5)+5 = 20
```

```python
# TODO: Another example - repeat operation
def repeat(func, num_times, value):
    result = value
    for _ in range(num_times):
        result = func(result)
    return result

def multiply_by_two(x):
    return x * 2

print(repeat(multiply_by_two, 3, 2))  # 2*2*2*2 = 16
```

## 3. Functions that Return Functions

We saw this with closures - returning a function allows for function factories and currying-like patterns.

```python
# TODO: Function returning another function
def create_adder(n):
    def adder(x):
        return x + n
    return adder

add_three = create_adder(3)
print(add_three(7))  # 10
```

## 4. Common Higher-Order Functions in Python

### 4.1 `map(function, iterable, ...)`

Applies function to each item. Returns an iterator.

```python
# TODO: map example
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
print(squared)
```

### 4.2 `filter(function, iterable)`

Filters items where function returns True.

```python
# TODO: filter example
numbers = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)
```

### 4.3 `reduce(function, iterable[, initializer])`

Applies function cumulatively to items. (From `functools` in Python 3)

```python
# TODO: reduce example
from functools import reduce
numbers = [1, 2, 3, 4, 5]
product = reduce(lambda x, y: x * y, numbers)
print(product)  # 120
```

### 4.4 `sorted(iterable, key=None, reverse=False)`

Takes `key` as a function - making it a HOF.

```python
# TODO: sorted with key
students = [('Alice', 90), ('Bob', 75), ('Charlie', 85)]
by_grade = sorted(students, key=lambda s: s[1])
print(by_grade)
```

### 4.5 `any()` and `all()`

Work well with predicates (functions returning bool).

```python
# TODO: any and all
numbers = [2, 4, 6, 8]
has_odd = any(x % 2 == 1 for x in numbers)
all_even = all(x % 2 == 0 for x in numbers)
print("Has odd:", has_odd)
print("All even:", all_even)
```

## 5. Function Composition

Function composition means combining simple functions to build more complex ones - applying one after another. In math, (f ∘ g)(x) = f(g(x)).

```python
# TODO: Manual function composition
def compose(f, g):
    def composed(x):
        return f(g(x))
    return composed

def double(x):
    return x * 2

def square(x):
    return x ** 2

# Apply square then double: double(square(x))
double_then_square = compose(double, square)
print(double_then_square(3))  # double(square(3)) = double(9) = 18
```

```python
# TODO: Flexible composition for multiple functions
def compose(*functions):
    def composed(x):
        result = x
        for func in reversed(functions):  # Apply from right to left
            result = func(result)
        return result
    return composed

def add_one(x): return x + 1
def double(x): return x * 2
def square(x): return x ** 2

f = compose(add_one, double, square)  # add_one(double(square(x)))
print(f(3))  # add_one(double(9)) = add_one(18) = 19
```

## 6. Practical Examples

### 6.1 Data Transformation Pipeline

```python
# TODO: Clean data pipeline with HOFs
users = [' alice ', 'BOB', '  charlie  ']
# Transform: trim and capitalize
clean = list(map(str.strip, users))
clean = list(map(str.capitalize, clean))
print(clean)

# Or in one go
clean2 = list(map(lambda s: s.strip().capitalize(), users))
print(clean2)
```

### 6.2 Grading System

```python
# TODO: Grade students with HOFs
scores = [95, 82, 67, 45, 88, 73]

# Get passing scores (>= 70)
passing = list(filter(lambda x: x >= 70, scores))
print("Passing:", passing)

# Calculate average of passing
from functools import reduce
avg_passing = reduce(lambda a, b: a + b, passing) / len(passing) if passing else 0
print("Avg passing:", avg_passing)
```

## 7. Imperative vs Declarative with HOFs

HOFs let us write more **declarative** code (what to do) rather than **imperative** (how to do it).

```python
# TODO: Imperative vs declarative
numbers = [1, 2, 3, 4, 5]

# Imperative approach
doubled_imp = []
for n in numbers:
    doubled_imp.append(n * 2)
print("Imperative:", doubled_imp)

# Declarative approach with HOF
doubled_dec = list(map(lambda x: x * 2, numbers))
print("Declarative:", doubled_dec)
```

## 9. Summary

- A **higher-order function** takes functions as arguments and/or returns functions.
- Functions are **first-class citizens** in Python.
- Common HOFs: `map()`, `filter()`, `reduce()`, `sorted()`, `any()`, `all()`.
- **Function composition** builds complex logic from simple reusable functions.
- HOFs enable **declarative programming** - focusing on *what* rather than *how*.
- They are fundamental to functional programming paradigms.
