## 1. What are Generators?

A **generator** is a special type of iterator that produces values on-the-fly using the `yield` keyword, rather than returning them all at once.

Key characteristics:
- Generators are **iterable** and follow the iterator protocol
- They use **lazy evaluation** - values are generated only when needed
- Much more **memory efficient** for large datasets
- A generator function pauses at `yield` and resumes from where it left off
- When a generator is exhausted, it raises `StopIteration`

### Generator Functions vs Regular Functions

| Feature | Regular Function | Generator Function |
| --- | --------------- | ----------------- |
| Returns | Single value (or None) | Generator object |
| Keyword | `return` | `yield` (can have multiple) |
| Execution | Runs to completion | Pauses and resumes |
| Memory | Stores all results | Generates values one at a time |
| State | Loses state after return | Preserves state between yields |

### Understanding `yield`

When the generator reaches `yield`:
1. It yields the value to the caller
2. It **pauses** execution and saves its state
3. When `next()` is called again, it **resumes** from where it left off

This is very different from `return`, which terminates the function completely.

## 5. Iterators vs Generators

While related, iterators and generators are not exactly the same.

| Aspect | Iterators | Generators |
| --- | --------- | ---------- |
| Definition | Any object implementing `__iter__()` and `__next__()` | A type of iterator created by generator functions/expressions |
| Creation | Manually implement the iterator protocol | Use `yield` or `()` generator expressions |
| State | You manage state manually | State is managed automatically |
| Boilerplate | More code required | Much simpler and cleaner |
| Memory | Depends on implementation | Inherently memory efficient |

**Note:** All generators are iterators, but not all iterators are generators.

## 9. Summary

- **Generators** are iterators that produce values lazily using `yield`.
- They **pause and resume** execution, preserving state between yields.
- **Memory efficient** - perfect for large or infinite sequences.
- **Generator expressions** `(... for x in ...)` are like list comprehensions but lazy.
- All generators are iterators; generators are simpler than custom iterator classes.
- Use `next()` to get values manually, or iterate with `for` loops.
- Generators enable powerful **streaming and pipeline** patterns.
